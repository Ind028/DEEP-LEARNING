```python
import os
path = kagglehub.dataset_download("uciml/adult-census-income")
print("Dataset path:", path)
df = pd.read_csv(os.path.join(path, "adult.csv")).replace(" ?", pd.NA).dropna()
for c in df.select_dtypes(include='object'):
    df[c] = LabelEncoder().fit_transform(df[c])
X = StandardScaler().fit_transform(df.drop("income", axis=1))
y = df["income"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)
model = Sequential([
    Dense(32, activation='relu', input_shape=(X_train.shape[1],)),
    Dense(16, activation='relu'),
    Dense(1, activation='sigmoid')
])
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model.fit(X_train, y_train, epochs=10, batch_size=32)
print(model.evaluate(X_test, y_test))
```

```python

path = kagglehub.dataset_download("uciml/indian-liver-patient-records")

print("Path to dataset files:", path)
```

```python
df = pd.read_csv(os.path.join(path, "indian_liver_patient.csv"))
df = df.dropna()
df["Gender"] = LabelEncoder().fit_transform(df["Gender"])
X = df.drop("Dataset", axis=1)
y = df["Dataset"]
X = StandardScaler().fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
model = Sequential([
    Dense(32, activation='relu', input_shape=(X_train.shape[1],)),
    Dense(16, activation='relu'),
    Dense(1, activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
model.fit(X_train, y_train, epochs=10, batch_size=32)
print(model.evaluate(X_test, y_test))
```

```python
path = kagglehub.dataset_download("heptapod/titanic")
print("Dataset path:", path)
print("Files in dataset:", os.listdir(path))
df = pd.read_csv(os.path.join(path, "train_and_test2.csv"))
df.columns = df.columns.str.strip()
df.rename(columns={
    "sibsp": "SibSp",
    "2urvived": "Survived"
}, inplace=True)
print("Columns:", df.columns)
df = df[["Pclass","Sex","Age","SibSp","Parch","Fare","Embarked","Survived"]]
df["Age"].fillna(df["Age"].median(), inplace=True)
df["Embarked"].fillna(df["Embarked"].mode()[0], inplace=True)
df["Sex"] = LabelEncoder().fit_transform(df["Sex"])
df["Embarked"] = LabelEncoder().fit_transform(df["Embarked"])
X = df.drop("Survived", axis=1)
y = df["Survived"]
X = StandardScaler().fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
model = Sequential([
    Dense(32, activation='relu', input_shape=(X_train.shape[1],)),
    Dense(16, activation='relu'),
    Dense(1, activation='sigmoid')
])
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
model.fit(X_train, y_train, epochs=10, batch_size=32)
loss, accuracy = model.evaluate(X_test, y_test)
print("Test Loss:", loss)
print("Test Accuracy:", accuracy)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler
np.random.seed(42)
rec=500
age=np.random.randint(20,65,rec)
family=np.random.randint(1,8,rec)
sal=np.random.randint(20000,150000,rec)
X=np.column_stack((age,family,sal))
y=((0.3*age+family*2+0.0005*sal)>50).astype(int)

scaler=StandardScaler()
X=scaler.fit_transform(X)
X_train, X_test,y_train,y_test= train_test_split(X,y,test_size=0.3,random_state=42)
#weight initialization
w=np.zeros(3)
b=0
lr=0.01
epochs=10

def sigmoid(inp,wt,b):
    z=np.dot(inp,wt)+b
    return 1/(1+np.exp(-z))

#before training plot
plt.scatter(X[:,0],X[:,1],c=y)
plt.title("Before training plt")
plt.show()
for _ in range(epochs):
    for i in range(len(X_train)):
        pred=sigmoid(X_train[i],w,b)
        error=y_train[i]-pred
        w=w+lr*error*X_train[i]
        b=b+lr*error

y_pred_full=[]
for sample in X_train:
    pred=sigmoid(sample,w,b)
    y_pred_full.append(pred)

plt.scatter(X[:,0],X[:,1],c=y)
plt.title("After training:")
plt.show()

print("weights",w)
print("biases:",b)


```
